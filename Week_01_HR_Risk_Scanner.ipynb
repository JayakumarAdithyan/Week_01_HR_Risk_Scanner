```python
def calculate_risk_score(emp):
    score = 0
    if emp["satisfaction_score"] < 0.4:
        score += 2
    if emp["overtime_hours"] >= 10:
        score += 2
    if emp["career_progression_score"] < 0.4:
        score += 1
    return score
```

```python
def risk_level(score):
    if score >= 4:
        return "CRITICAL"
    elif score >= 3:
        return "HIGH"
    elif score >= 1:
        return "MEDIUM"
    return "LOW"
```

```python
print(calculate_risk_score(employees[0]))
print(risk_level(calculate_risk_score(employees[0])))
```

```python
results = []
for i, emp in enumerate(employees):
    score = calculate_risk_score(emp)
    level = risk_level(score)
    results.append((emp["department"], level, emp["left_company"]))
    print(i, emp["department"], score, level)
```

```python
for lvl in ["LOW", "MEDIUM", "HIGH", "CRITICAL"]:
    count = 0
    for dept, level, left in results:
        if level == lvl:
            count += 1
    print(lvl, count)
```

```python
bad_emp = {"department": "Sales", "satisfaction_score": "abc", "overtime_hours": 12, "career_progression_score": 0.3, "left_company": 0}
print(calculate_risk_score(bad_emp))
```

```python
test_data = employees + [bad_emp]

results = []
skipped = 0
for emp in test_data:
    try:
        score = calculate_risk_score(emp)
        level = risk_level(score)
        results.append((emp["department"], level, emp["left_company"]))
    except (KeyError, TypeError, ValueError):
        skipped += 1

print("Scored:", len(results))
print("Skipped:", skipped)
```

```python
counts = {}
for dept, level, left in results:
    if dept not in counts:
        counts[dept] = {}
    counts[dept][level] = counts[dept].get(level, 0) + 1

for dept, c in counts.items():
    print(dept, c)
```

```python
employees = list(ds.take(1000))
print(len(employees))
```

```python
results = []
skipped = 0
for emp in employees:
    try:
        score = calculate_risk_score(emp)
        level = risk_level(score)
        results.append((emp["department"], level, emp["left_company"]))
    except (KeyError, TypeError, ValueError):
        skipped += 1

print("Scored:", len(results))
print("Skipped:", skipped)
```

```python
for lvl in ["LOW", "MEDIUM", "HIGH", "CRITICAL"]:
    total = 0
    left_count = 0
    for dept, level, left in results:
        if level == lvl:
            total += 1
            left_count += left
    if total > 0:
        print(lvl, total, "employees,", round(100 * left_count / total, 1), "% left")
```
